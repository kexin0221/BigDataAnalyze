```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
```

```python
driver_path = ChromeDriverManager().install()
service = Service(driver_path)
```

```python
browser = webdriver.Chrome(service=service)
browser.get('https://weibo.com/1699432410/QzJ1YekFd')
time.sleep(5)
```

```python
processed = set()
for _ in range(20):
    browser.execute_script("window.scrollBy(0, 1000);")
    time.sleep(2)

    # 重点：使用评论区特有的类名 m-detail-main
    elements = browser.find_elements(By.CSS_SELECTOR, ".wbpro-list .wbpro-feed-content")
    for el in elements:
        txt = el.text.replace('\n', ' ').strip()
        if txt and txt not in processed:
            print(txt)
            processed.add(txt)
```

```python
browser.quit()
```
