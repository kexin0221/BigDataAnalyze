```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
```

```python
driver_path = ChromeDriverManager().install()
service = Service(driver_path)
```

```python
browser = webdriver.Chrome(service=service)
url = 'https://www.sse.com.cn/disclosure/credibility/supervision/inquiries/'
browser.get(url)
time.sleep(3) # 这里必须要加3秒的延迟，因为有个刷新的动作需要等待下
data = browser.page_source
data
```

```python
import re
p_title='<a class="table_titlewrap" href=".*?" target="_blank">(.*?)</a>'
p_href='<a class="table_titlewrap" href="(.*?)" target="_blank">'
title=re.findall(p_title, data,re.S)
href=re.findall(p_href, data,re.S)
title, href
```
