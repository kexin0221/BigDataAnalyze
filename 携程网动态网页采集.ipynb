```python
import time

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
```

```python
driver_path = ChromeDriverManager().install()
service = Service(driver_path)
browser = webdriver.Chrome(service=service)
```

```python
url = 'https://hotels.ctrip.com/hotels/list?countryId=1&city=12&provinceId=0&checkin=2026/04/15&checkout=2026/04/16&optionId=12&optionType=City&display=%E5%8D%97%E4%BA%AC'
browser.get(url)
time.sleep(5)
```

```python
hotel_names = set()
while len(hotel_names) < 100:
    elements = browser.find_elements(By.CLASS_NAME, 'hotelName')
    for el in elements:
        name = el.text
        if name and name not in hotel_names:
            hotel_names.add(name)
            print(f"{len(hotel_names)}: {name}")
            if len(hotel_names) >= 100: break

    browser.execute_script("window.scrollBy(0, 2000);")
    time.sleep(2)
```

```python
browser.quit()
```
