```python
browser = webdriver.Chrome(service=service)
url = 'http://www.sse.com.cn/disclosure/credibility/supervision/inquiries/'
browser.get(url)
time.sleep(3) # 这里必须要加3秒的延迟，因为有个刷新的动作需要等待下
data = browser.page_source
print(data)
```

# 批量下载PDF

```python
for i in range(len(href)):
    import requests
    res = requests.get(href[i])
    path = 'D:/code/python/BigDataAnalyze/上交所问询函/' + title[i]+ '.pdf'
    file = open(path, 'wb')
    file.write(res.content)
    file.close()
```

# 获取多页信息

```python

url = 'http://www.sse.com.cn/disclosure/credibility/supervision/inquiries/'
browser.get(url)
time.sleep(3)  # 这里必须要加3秒的延迟，因为有个刷新的动作需要等待下

data_all = ''  # 创建一个空字符串，用来汇总之后各网页的网页源代码

for i in range(10):
    browser.find_element(By.XPATH,'/html/body/div[9]/div/div[2]/div/div[1]/div[2]/span[1]/input').send_keys(i+1)
    browser.find_element(By.XPATH,'/html/body/div[9]/div/div[2]/div/div[1]/div[2]/span[2]/a').click()
    time.sleep(3)  # 这里必须要加3秒的延迟，因为有个刷新的动作需要等待下
    data = browser.page_source
    data_all = data_all + data  # 也可以简写为data_all += data
```

```python
import re
p_title='<a class="table_titlewrap" href=".*?" target="_blank">(.*?)</a>'
p_href='<a class="table_titlewrap" href="(.*?)" target="_blank">'
title=re.findall(p_title,data_all,re.S)
href=re.findall(p_href, data_all,re.S)


title, href, len(title)
```
